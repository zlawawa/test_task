```python
import pandas as pd
```

```python
df = pd.read_excel('analytics_report_2026-03-30_10_15.xlsx', skiprows = 4)
```

```python
df.head(3)
```

```python
df = df[1:]
```

```python
df.head(3)
```

```python
df = df.loc[:, ['Название товара', 'Бренд', 'Продавец', 'Признак товара', 'Заказано на сумму, ₽', 'Динамика оборота, %', 'Доля выкупа, %', 'Доля оборота в акциях, %']]
```

```python
df.dtypes
```

```python
top_brands = df.groupby('Бренд')['Заказано на сумму, ₽'].sum()
```

```python
top_brands.head(10)
```

```python
all_sum = df['Заказано на сумму, ₽'].sum()
```

```python
all_sum
```

```python
result = (top_brands.head(10) / all_sum) * 100
```

```python
result
```

```python
result_rename_column = result.reset_index(name = 'процент от оборота')
```

```python
result_rename_column.to_excel('result_task_one.xlsx', index = False)
```
